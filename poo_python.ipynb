## O que é POO?

- Paradigma que organiza o código em **objetos**, instâncias de **classes**.
- Foca em **abstração**, **encapsulamento**, **herança** e **polimorfismo**.
- Python oferece POO de forma **flexível** e **intuitiva**.

### Por que usar POO?

- **Reutilização** (herança, composição)
- **Organização** (modularidade)
- **Manutenção** (encapsulamento)
- **Modelagem** mais próxima do mundo real

### Vocabulário

| Termo | Definição |

| Classe | Molde com atributos e métodos |
| Objeto | Instância de uma classe |
| Atributo | Dado do objeto (estado) |
| Método | Comportamento do objeto (função) |

## Definindo classes

```python
class Pessoa:
    def __init__(self, nome, idade):  # construtor
        self.nome = nome      # atributo de instância (público)
        self.__idade = idade  # "privado" (name mangling)

    def apresentar(self):     # método de instância
        print(f"Olá, eu sou {self.nome}!")
```

### Sobre `self`

- Primeiro parâmetro dos métodos de instância.
- Referencia o *objeto atual*.
- Nome é convenção; poderia ser outro, mas `self` é padrão.

### Encapsulamento

- `_protegido` (convenção)
- `__privado` (name mangling: `_Classe__nome`)

```python
p1 = Pessoa("Alice", 30)
p1.apresentar()  # Olá, eu sou Alice!
p2 = Pessoa("Bob", 25)
p2.apresentar()  # Olá, eu sou Bob!
```

**Checando os dados de uma Classe**

```python
class Pessoa:
    def __init__(self, nome, idade, sexo):  # construtor
        if len(nome) < 2:
            raise ValueError("Nome deve ter pelo menos 3 caracteres!")
        if idade < 0:
            raise ValueError("Idade inválida!")
        if sexo not in ['M', 'F']:
            raise ValueError("Sexo deve ser 'M' ou 'F'!")
        self.nome = nome      # atributo de instância (público)
        self.sexo = sexo      # atributo de instância (público)
        self.__idade = idade  # "privado" (name mangling)

    def apresentar(self):     # método de instância
        print(f"Olá, eu sou {self.nome} e tenho {self.__idade} anos!")
        
    def get_idade(self):  # método getter para acessar o atributo "privado"
        return self.__idade
    
    def set_idade(self, idade):  # método setter para modificar o atributo "privado"
        if idade >= 0:
            self.__idade = idade
        else:
            raise ValueError("Idade inválida!")
```

```python
p1 = Pessoa("Alice", 30, "F")
p1.apresentar()  # Olá, eu sou Alice!
p2 = Pessoa("Bob", 25, "M")
p2.apresentar()  # Olá, eu sou Bob!
```

### Métodos estáticos e de classe

```python
class Calculadora:
    @staticmethod
    def soma(a, b):
        return a + b
    
    @staticmethod
    def subtracao(a, b):
        return a - b
    
    @staticmethod
    def multiplicacao(a, b):
        return a * b
    
    @staticmethod
    def divisao(a, b):
        if b == 0:
            raise ValueError("Divisão por zero não é permitida!")
        return a / b

    @classmethod
    def add_from_str(cls, string):
        a, b = map(int, string.split(","))
        return cls.soma(a, b)

    @classmethod
    def minus_from_str(cls, string):
        a, b = map(int, string.split(","))
        return cls.subtracao(a, b)
    
    @classmethod
    def multiply_from_str(cls, string):
        a, b = map(int, string.split(","))
        return cls.multiplicacao(a, b)
    
    @classmethod
    def divide_from_str(cls, string):
        a, b = map(int, string.split(","))
        return cls.divisao(a, b)
```

```python
print(f"Soma de 2 e 3: {Calculadora.soma(2,3)}")     # 5
print(f"Soma 4,6: {Calculadora.add_from_str("4,6")}")  # 10
print(f"Subtração 10,4: {Calculadora.minus_from_str("10,4")}")  # 6
print(f"Multiplicação 3,5: {Calculadora.multiply_from_str("3,5")}")  # 15
try:
    print(f"Divisão 10,0: {Calculadora.divide_from_str("10,0")}")  # Deve levantar um ValueError
except ValueError as e:
    print(e)  # Divisão por zero não é permitida!
```

## Propriedades (`@property`)

```python
class Conta:
    def __init__(self, saldo=0.0):
        self._saldo = saldo

    @property
    def saldo(self):         # getter
        return self._saldo

    @saldo.setter
    def saldo(self, valor):  # setter com validação
        if valor < 0:
            raise ValueError("Saldo não pode ser negativo")
        self._saldo = valor
```

```python
# 1. Criando uma instância da classe Conta
minha_conta = Conta(100.0)

# ---------------------------------------------------------
# USANDO O GETTER (Lendo o valor)
# ---------------------------------------------------------
# Note que acessamos 'saldo' sem os parênteses, como se fosse um atributo normal.
print(f"Saldo inicial: R$ {minha_conta.saldo:.2f}")

# ---------------------------------------------------------
# USANDO O SETTER (Alterando o valor com sucesso)
# ---------------------------------------------------------
# Ao atribuir um valor, o Python chama o método setter automaticamente.
minha_conta.saldo = 250.50
print(f"Saldo após depósito: R$ {minha_conta.saldo:.2f}")

# ---------------------------------------------------------
# TESTANDO A VALIDAÇÃO DO SETTER (Tentando colocar valor negativo)
# ---------------------------------------------------------
# Como o setter não permite valores negativos, ele lançará um ValueError.
# Usamos try/except para capturar o erro e o programa não quebrar.
try:
    print("\nTentando definir um saldo negativo...")
    minha_conta.saldo = -50.0
except ValueError as erro:
    print(f"Operação bloqueada! Motivo: {erro}")

# Mostrando que o saldo não foi alterado pela tentativa falha
print(f"Saldo final (inalterado): R$ {minha_conta.saldo:.2f}")
```

## Métodos especiais (dunder)

```python
class Vetor2D:
    def __init__(self, x, y): 
        self.x, self.y = x, y
    
    def __repr__(self): # Define como o objeto deve ser "visto" quando impresso ou inspecionado.
        return f"Vetor2D({self.x}, {self.y})"
    
    def __add__(self, other): #  Sobrecarrega o operador de soma (+)
        return Vetor2D(self.x+other.x, self.y+other.y)
    
    def __eq__(self, other): # Sobrecarrega o operador de igualdade (==)
        return self.x==other.x and self.y==other.y

```

```python
a = Vetor2D(1,2)
b = Vetor2D(3,4)
print(a + b)  # Vetor2D(4, 6)
```

## Herança

```python
class Pessoa:
    def __init__(self, nome): 
        self.nome = nome
    def apresentar(self): print(f"Olá, sou {self.nome}.")

class Aluno(Pessoa):
    def __init__(self, nome, matricula):
        super().__init__(nome)
        self.matricula = matricula

    def apresentar(self):
        super().apresentar()
        print(f"Matrícula: {self.matricula}")
```

```python
# ---------------------------------------------------------
# 1. USANDO A CLASSE BASE (Pessoa)
# ---------------------------------------------------------
pessoa_comum = Pessoa("João Silva")
pessoa_comum.apresentar()

print("-" * 40) # Apenas um separador visual

# ---------------------------------------------------------
# 2. USANDO A CLASSE FILHA (Aluno)
# ---------------------------------------------------------
aluno = Aluno("Maria Souza", "2023001")
aluno.apresentar()

print("-" * 40)

# ---------------------------------------------------------
# 3. A "MÁGICA" DO POLIMORFISMO (Dica de Ouro)
# ---------------------------------------------------------
# Como ambas as classes têm o método apresentar(), 
# podemos misturá-las em uma lista e tratar todas igualmente.

grupo = [
    Pessoa("Carlos (Visitante)"),
    Aluno("Ana", "998877"),
    Pessoa("Pedro (Funcionário)"),
    Aluno("Bruno", "112233")
]

print("\nApresentação da turma:")
for individuo in grupo:
    individuo.apresentar() # O Python sabe qual método chamar!
    print("---")
```

## Polimorfismo

```python
class Forma:
    def area(self): 
        raise NotImplementedError

class Circulo(Forma):
    def __init__(self, r): 
        self.r = r
    def area(self): 
        return 3.14159 * self.r**2

class Quadrado(Forma):
    def __init__(self, l): 
        self.l = l
    def area(self): 
        return self.l**2

# --- ADICIONANDO UMA NOVA FORMA (Para provar o polimorfismo) ---
class Retangulo(Forma):
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura
    def area(self):
        return self.base * self.altura
    
def total_area(formas):
    return sum(f.area() for f in formas)
```

```python
# ---------------------------------------------------------
# EXEMPLO DE USO (A Mágica do Polimorfismo)
# ---------------------------------------------------------

# 1. Criando uma lista com objetos de tipos DIFERENTES
formas_geometricas = [
    Circulo(5),          # Raio 5
    Quadrado(4),         # Lado 4
    Retangulo(3, 6),     # Base 3, Altura 6 (Adicionado depois!)
    Circulo(2)           # Raio 2
]

# 2. Iterando sobre a lista (Polimorfismo em ação)
print("--- Áreas Individuais ---")
for forma in formas_geometricas:
    # type(forma).__name__ pega o nome da classe dinamicamente
    nome_classe = type(forma).__name__ 
    print(f"A área do {nome_classe} é: {forma.area():.2f}")

# 3. Usando a função que trata todos igualmente
print("\n--- Cálculo Total ---")
soma_total = total_area(formas_geometricas)
print(f"A área total de todas as formas combinadas é: {soma_total:.2f}")
```

### Classes abstratas (ABC)

```python
from abc import ABC, abstractmethod

# ---------------------------------------------------------
# 1. O CONTRATO (Sua Classe Abstrata)
# ---------------------------------------------------------
class Repositorio(ABC):
    @abstractmethod
    def salvar(self, obj): ...
    
    @abstractmethod
    def buscar(self, id): ...

# ---------------------------------------------------------
# 2. MODELO DE DADOS (O que vamos salvar)
# ---------------------------------------------------------
class Usuario:
    def __init__(self, id, nome):
        self.id = id
        self.nome = nome
        
    def __repr__(self):
        return f"Usuario(id={self.id}, nome='{self.nome}')"

# ---------------------------------------------------------
# 3. IMPLEMENTAÇÕES CONCRETAS (Respeitando o contrato)
# ---------------------------------------------------------

# Implementação 1: Salvando na memória (RAM)
class RepositorioEmMemoria(Repositorio):
    def __init__(self):
        self._dados = {}  # Dicionário para guardar os objetos
        
    def salvar(self, obj):
        print(f"[Memória] Salvando {obj}...")
        self._dados[obj.id] = obj
        
    def buscar(self, id):
        print(f"[Memória] Buscando usuário com ID {id}...")
        return self._dados.get(id)

# Implementação 2: Salvando em um "Banco de Dados" (Simulado)
class RepositorioBancoDados(Repositorio):
    def __init__(self):
        self._tabela = [] # Lista simulando uma tabela SQL
        
    def salvar(self, obj):
        print(f"[Banco de Dados] Executando INSERT INTO usuarios...")
        self._tabela.append(obj)
        
    def buscar(self, id):
        print(f"[Banco de Dados] Executando SELECT * FROM usuarios WHERE id={id}...")
        for usuario in self._tabela:
            if usuario.id == id:
                return usuario
        return None
```

```python
# Criando os repositórios
repo_memoria = RepositorioEmMemoria()
repo_banco = RepositorioBancoDados()
    
# Criando um usuário
u1 = Usuario(1, "Alice")
u2 = Usuario(2, "Bob")
    
# Usando o Repositório em Memória
print("--- Testando Repositório em Memória ---")
repo_memoria.salvar(u1)
print(f"Busca: {repo_memoria.buscar(1)}\n")
    
# Usando o Repositório de Banco de Dados
print("--- Testando Repositório de Banco de Dados ---")
repo_banco.salvar(u2)
print(f"Busca: {repo_banco.buscar(2)}\n")

# A MÁGICA DO POLIMORFISMO:
# Funções podem aceitar a classe abstrata como parâmetro (Type Hinting)
def processar_usuario(repo: Repositorio, user_id: int):
    usuario = repo.buscar(user_id)
    if usuario:
        print(f"Processando usuário: {usuario.nome}")
    else:
        print("Usuário não encontrado.")
            
print("--- Polimorfismo em Ação ---")
processar_usuario(repo_memoria, 1)
processar_usuario(repo_banco, 2)
```

### MRO & herança múltipla (cuidado)

## Composição (preferida)

- Objetos **contêm** outros objetos para delegar responsabilidades.
- Reduz acoplamento e tende a ser mais simples de manter.

### Exemplo

```python
# ---------------------------------------------------------
# 1. OS COMPONENTES (As implementações de Logger)
# ---------------------------------------------------------

# O Logger original (Imprime no console)
class Logger:
    def log(self, msg): 
        print(f"[LOG CONSOLE] {msg}")

# Um novo Logger (Simula salvar em um arquivo de texto)
class LoggerArquivo:
    def log(self, msg): 
        # Na vida real, aqui abriria um arquivo e escreveria nele
        print(f"[LOG ARQUIVO] Gravando em disco: {msg}")


# ---------------------------------------------------------
# 2. A CLASSE QUE USA A COMPOSIÇÃO
# ---------------------------------------------------------
class ServicoEmail:
    # Injeção de Dependência: O logger é passado de fora
    def __init__(self, logger): 
        self.logger = logger  # <-- AQUI ACONTECE A COMPOSIÇÃO
        
    def enviar(self, para, msg):
        self.logger.log(f"Iniciando envio para {para}")
        
        # ... Lógica complexa de envio de email ...
        print(f"   -> Email entregue com sucesso para {para}!")
        
        self.logger.log("Processo de envio finalizado (OK)")
```

```python
# ---------------------------------------------------------
# 3. EXEMPLO DE USO
# ---------------------------------------------------------
print("="*50)
print("Cenário 1: Usando o Logger padrão (Console)")
print("="*50)
    
# Criamos o serviço "injetando" o Logger normal
servico_comum = ServicoEmail(Logger())
servico_comum.enviar("alice@email.com", "Bem-vinda!")
    
print("\n" + "="*50)
print("Cenário 2: Usando um Logger diferente (Arquivo)")
print("="*50)
    
# Sem alterar NENHUMA linha na classe ServicoEmail, 
# trocamos o comportamento apenas passando outro objeto!
servico_producao = ServicoEmail(LoggerArquivo())
servico_producao.enviar("bob@email.com", "Sua senha foi resetada.")
```

## Caso de uso (mini-sistema)

Cadastro e persistência simples em arquivo (POO + I/O).

```python
class Pessoa:
    def __init__(self, cpf, nome, telefone):
        self.cpf, self.nome, self.telefone = cpf, nome, telefone
    def __str__(self):
        return f"{self.cpf},{self.nome},{self.telefone}"

class GerenciadorPessoas:
    def __init__(self, arquivo="pessoas.txt"):
        self.pessoas = {}    # dict[cpf] = Pessoa
        self.arquivo = arquivo
        self.carregar()

    def cadastrar(self, cpf, nome, telefone):
        self.pessoas[cpf] = Pessoa(cpf, nome, telefone)

    def listar(self):
        return list(self.pessoas.values())

    def salvar(self):
        with open(self.arquivo, "w", encoding="utf-8") as f:
            for p in self.pessoas.values():
                f.write(str(p) + "\n")  # Ajustado para \n (pular linha)

    def carregar(self):
        try:
            with open(self.arquivo, "r", encoding="utf-8") as f:
                for linha in f:
                    cpf, nome, tel = linha.strip().split(",")
                    self.pessoas[cpf] = Pessoa(cpf, nome, tel)
        except FileNotFoundError:
            pass # Se o arquivo não existe, começa com o dicionário vazio


```

```python
nome_arquivo = "agenda_contatos.txt"

print("=" * 50)
print("1. INICIANDO PROGRAMA E CADASTRANDO PESSOAS")
print("=" * 50)
    
# Como o arquivo ainda não existe, o carregar() não faz nada (passa direto)
agenda = GerenciadorPessoas(nome_arquivo)
    
agenda.cadastrar("111.222.333-44", "Ana Silva", "(11) 98888-7777")
agenda.cadastrar("555.666.777-88", "Bruno Costa", "(21) 97777-6666")
agenda.cadastrar("999.000.111-22", "Carla Dias", "(31) 96666-5555")
    
print("Pessoas em memória:")
for pessoa in agenda.listar():
    print(f" - {pessoa.nome} | Tel: {pessoa.telefone}")


print("\n" + "=" * 50)
print("2. SALVANDO NO ARQUIVO DE TEXTO")
print("=" * 50)
agenda.salvar()
print(f"Dados salvos com sucesso em '{nome_arquivo}'!")

print("\n" + "=" * 50)
print("3. SIMULANDO REINÍCIO DO PROGRAMA (Nova Instância)")
print("=" * 50)

# Agora criamos um NOVO gerenciador. 
# O __init__ vai chamar o carregar() e ler o arquivo que acabamos de salvar!
nova_agenda = GerenciadorPessoas(nome_arquivo)
    
print("Dados recuperados do arquivo:")
for pessoa in nova_agenda.listar():
    print(f" - CPF: {pessoa.cpf} | Nome: {pessoa.nome} | Tel: {pessoa.telefone}")
```

## Boas práticas em POO (Python)

- Use nomes claros e docstrings em classes/métodos.
- Evite herança desnecessária; prefira composição.
- Encapsule com propriedades para validação.
- Pequenos métodos coesos; aplique SRP/SOLID.
- Use `dataclasses` para objetos de dados.
- Adote **type hints** para legibilidade e ferramentas.

## Cheat Sheet
